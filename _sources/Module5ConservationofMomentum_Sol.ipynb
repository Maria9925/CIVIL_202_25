**Solution**

````python
img = Image.open(r"figures/25_Momentum_1.png")
img= img.resize((300,190))
img
````

```{Dropdown} Show code cell outputs

Continuity equation:

$$
Q_1 &= Q_2 \\
v_1 A_1 &= v_2 A_2 \\
v_1(y_1 b) &= v_2(y_2 b) \\
v_1 \frac{y_1}{y_2} &= v_2 \\
$$

Conservation of energy:

$$
z_1 + \frac{p_1}{\rho g} + \frac{v_1^2}{2g} &= z_2 + \frac{p_2}{\rho g} + \frac{v_2^2}{2g} + h_f \\
\frac{p_1}{\rho g} &= y_1, \quad \frac{p_2}{\rho g} = y_2 \\
z_1 = z_2 \\
y_1 + \frac{v_1^2}{2g} &= y_2 + \frac{y_1^2 v_1^2}{2g y_2^2} + h_f
$$

Solving for $( v_1)$:

$$
v_1 = \left( \frac{2g(y_2 - y_1 + h_f)}{1 - \frac {y_1^1}{y_2^2}} \right)^{0.5}
$$

$N \mathrm{II}$:

$$
\sum F_x &= \rho Q (v_{2x} - v_{1x}) \\
F_1 - F_2 - R_x &= \rho Q (v_{2x} - v_{1x})
$$
```

````python
%%render long 
h_f=0.2 #m
y_1=2.2 #m
y_2=0.4 #m
y_3=0.5 #m
b=1 #m


v_1= ((2*g*(y_2-y_1+h_f))/(1-((y_1**2)/(y_2**2))))**0.5 #m/s
v_2= 5.5*v_1 #m/s
A_1= y_1*b #m2
A_2= y_2*b #m2
Q= v_1*A_1 #m3/s
R_x= rho_water*g*y_1/2*y_1-rho_water*g*y_2/2*y_2-rho_water*Q*(v_2-v_1) #N
R_x_=R_x/1000 #kN
````

```{Dropdown} Show code cell outputs
The force on the structure is the same size but in the opposite direction ($N \mathrm{III}$）:
```

````python
%%render 2
F=R_x_ #kN to the right
````

```{Dropdown} Show code cell outputs

Conservation of mass:

$$
Q_1 &= Q_2 \\
v_1 A_1 &= v_2 A_2 \\
v_1 y_1 b &= v_2 y_2 b \\
v_2 &= \frac{y_1 v_1}{y_2}
$$

Conservation of Energy:

$$
z_1 + \frac{P_1}{\rho g} + \frac{v_1^2}{2g} &= z_2 + \frac{P_2}{\rho g} + \frac{v_2^2}{2g} + h_f \\
\frac{P_1}{\rho g} &= y_1, \quad \frac{P_2}{\rho g} = y_2\\
z_1 = z_2\\
$$


$$
\frac{v_1^2}{2g} &= y_2 + \frac{v_2^2}{2g} + h_f - y_1 \\
\frac{v_1^2}{2g} &= y_2 + \frac{y_1^2 v_1^2}{2g y_2^2} + h_f - y_1
$$

Once the velocities at both sections are determined, the conservation of momentum can then be applied:

$$
\sum F_x &= \rho Q (v_2 - v_1) \\
F_1 - F_2 -R_x &= \rho Q (v_2 - v_1)
$$

The centroid of the spillway is:

$$
\rho g \frac{y_1}{2} A_1 - \rho g \frac{y_2}{2} A_2 - F = \rho Q (v_2 - v_1)
$$


```

````python
%%render long 
b= 100#m 
y_1= 20 #m 
y_2= 1.5 #m 
h_f= 1.2 #m 


v_1= (((y_1-y_2-h_f)*2*g)/((y_1/y_2)**2-1))**0.5 #m/s
v_2= y_1*v_1/y_2#m/s
Q= v_1*y_1*b #m3/s
A_1= b*y_1 #m2
A_2= b*y_2 #m2
R_x= rho_water*g*y_1*A_1/2-rho_water*g*y_2*A_2/2-rho_water*Q*(v_2-v_1) #N
R_x= R_x/1000000 #MN
````

$$

$$
z_1+\frac {P_1}{\rho g}+\frac{v_1^2}{2g}=z_2+\frac {P_2}{\rho g}+\frac{v_2^2}{2g}\\

````python
%%render long 
v_2= v_1*(y_1*b_1)/(y_2*b_2) #m/s
Q= v_1*(y_1*b_1) #m3/s
A_1= b_1*y_1 #m2
A_2= b_2*y_2 #m2
R_x= rho_water*g*y_1/2*A_1-rho_water*g*y_2/2*A_2-rho_water*Q*(v_2-v_1) #N
R_x_= R_x/1000 #kN
````

A perforated plate with 50% of its area consisting of small holes is installed at the outlet of a 30 cm diameter horizontal water pipe carrying water at a velocity of 2 m/s. Use conservation of mass to find the water velocity through the holes.

**Solution**

```{Dropdown} Show code cell outputs
The first step is calculate the velocity in section 2 by using the conservation of mass: 

$$
Q_1=Q_2\\
v_1*A_1=v_2*A_2\\
$$

where $A_2 = 0.5*A_1$ according the question.  

$$
v_1*A_1=v_2*0.5*A_1\\
v_2=\frac {v_2}{0.5}
$$

Once the velocity is calculated, the pressure in section one can be calculated with the Bernoulli equation:

$$
z_1+\frac {P_1}{\rho g}+\frac{v_1^2}{2g}=z_2+\frac {P_2}{\rho g}+\frac{v_2^2}{2g}\\
$$

where $z_1= z_2$, and $P_2= 0 Pa$

$$
\frac {P_1}{\rho g}=\frac{v_2^2}{2g}-\frac{v_1^2}{2g}
$$

Once the pressure in section one is calculated, the force of the water can be calculated with the conservation of Momentum:

$$
\sum F_x = \rho Q (v_2-v_1)\\
P_1A_1-P_2A_2-R_x = \rho Q (v_2-v_1)\\
$$

where $P_2$ is atmospheric pressure= 0 

$$
R_x = -\rho Q (v_2-v_1)+P_1A_1\\
$$
```

````python
%%render long 
v_1= 2#m/s
D_1= 30/100 #m

v_2 = v_1/0.5 #m/s
A_1= pi*D_1**2/4
Q= v_1*A_1  #m3/s
P_1= (v_2**2/(2*g)-v_1**2/(2*g))*g*rho_water #Pa
R_x= P_1*A_1-rho_water*Q*(v_2-v_1) #N
````

$$

In the y-axis:
